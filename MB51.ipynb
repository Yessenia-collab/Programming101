### Identificación de Materiales con Crecimiento en Movimientos

```python
import pandas as pd

# Asegurar que la columna 'Posting Date' esté en formato de fecha
df_filtered['Posting Date'] = pd.to_datetime(df_filtered['Posting Date'], errors='coerce')
df_filtered.dropna(subset=['Posting Date'], inplace=True)

# Extraer el año de la 'Posting Date'
df_filtered['Year'] = df_filtered['Posting Date'].dt.year

# Definir los años de interés
years_of_interest = [2024, 2025, 2026]

# Filtrar el DataFrame para incluir solo los años de interés
df_yearly_quantity = df_filtered[df_filtered['Year'].isin(years_of_interest)]

# Agrupar por Material y Año, y sumar la Quantity
material_yearly_sum = df_yearly_quantity.groupby(['Material', 'Year'])['Quantity'].sum().reset_index()

# Pivotar la tabla para tener los años como columnas
material_yearly_summary_pivot = material_yearly_sum.pivot_table(
    index='Material',
    columns='Year',
    values='Quantity'
).fillna(0)

# Renombrar las columnas para mayor claridad
material_yearly_summary_pivot.columns = [f'Quantity_{year}' for year in material_yearly_summary_pivot.columns]

print("Suma de 'Quantity' por Material y Año (2024, 2025, 2026):")
display(material_yearly_summary_pivot.head())
```

```python
import pandas as pd

# Asumiendo que material_yearly_summary_pivot está disponible de la celda anterior.

# Definir los nombres de las columnas para los años
col_2024 = 'Quantity_2024'
col_2025 = 'Quantity_2025'
col_2026 = 'Quantity_2026'

# Asegurar que las columnas existen en el DataFrame, si no, se asumen 0 para los cálculos.
# Esto maneja casos donde no hay datos para un año específico pero queremos hacer la comparación.
for col in [col_2024, col_2025, col_2026]:
    if col not in material_yearly_summary_pivot.columns:
        material_yearly_summary_pivot[col] = 0.0

# Identificar materiales que mostraron un incremento de 2024 a 2025
# Esto incluye materiales que tuvieron 0 en 2024 y > 0 en 2025
increased_24_25 = (material_yearly_summary_pivot[col_2025] > material_yearly_summary_pivot[col_2024])

# Identificar materiales que mostraron un incremento de 2025 a 2026
# Esto incluye materiales que tuvieron 0 en 2025 y > 0 en 2026
increased_25_26 = (material_yearly_summary_pivot[col_2026] > material_yearly_summary_pivot[col_2025])

# Un material se considera que ha 'incrementado' si hubo un aumento en la cantidad
# de 2024 a 2025 O de 2025 a 2026.
materials_with_increment = material_yearly_summary_pivot[increased_24_25 | increased_25_26]

print("Materiales con 'Quantity' incrementada en al menos un período (2024-2026):")
display(materials_with_increment.head())
print(f"\nTotal de materiales con incremento: {len(materials_with_increment)}")

# --- Comienza la sección para el análisis de Pareto ---

# Calcular el incremento neto de 2024 a 2026
# Si un material no tiene datos para 2024 o 2026, ya se rellenó con 0 en los pasos anteriores.
materials_with_increment['Net_Increase_24_26'] = materials_with_increment[col_2026] - materials_with_increment[col_2024]

# Filtrar solo los materiales con un incremento neto positivo para el análisis de Pareto
pareto_df = materials_with_increment[materials_with_increment['Net_Increase_24_26'] > 0].copy()

# Ordenar por el incremento neto de forma descendente
pareto_df = pareto_df.sort_values(by='Net_Increase_24_26', ascending=False)

# Calcular el porcentaje de cada incremento respecto al total de incrementos netos
total_net_increase = pareto_df['Net_Increase_24_26'].sum()
pareto_df['Percentage_of_Increase'] = (pareto_df['Net_Increase_24_26'] / total_net_increase) * 100

# Calcular el porcentaje acumulado
pareto_df['Cumulative_Percentage_of_Increase'] = pareto_df['Percentage_of_Increase'].cumsum()

print("\nAnálisis de Pareto del mayor incremento (Neto 2024-2026):")
display(pareto_df.head(10)) # Mostrar los top 10 materiales en el análisis de Pareto
```
